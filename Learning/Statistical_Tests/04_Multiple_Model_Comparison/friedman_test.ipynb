```python
import numpy as np
from scipy.stats import friedmanchisquare
```

```python
tinyllama = [0.70, 0.65, 0.72, 0.68, 0.75]

smollm2   = [0.75, 0.78, 0.76, 0.77, 0.79]

qwen      = [0.85, 0.80, 0.88, 0.83, 0.87]

phi3      = [0.88, 0.86, 0.89, 0.87, 0.90]
```

```python
stat, p_value = friedmanchisquare(
    tinyllama,
    smollm2,
    qwen,
    phi3
)

print("Friedman statistic:", stat)
print("p-value:", p_value)
```

```python
if p_value < 0.05:
    print("Reject null hypothesis - there is a significant difference")
else:
    print("Fail to reject null hypothesis - no significant difference")
```
