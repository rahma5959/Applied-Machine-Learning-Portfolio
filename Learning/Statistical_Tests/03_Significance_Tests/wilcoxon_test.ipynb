```python
import numpy as np
from scipy.stats import wilcoxon
```

```python
tinyllama_scores = [0.70, 0.65, 0.72, 0.68, 0.75]
qwen_scores      = [0.85, 0.80, 0.88, 0.83, 0.87]
```

```python
stats,p_value = wilcoxon(tinyllama_scores, qwen_scores)
print(f"Wilcoxon test: statistic={stats:.4f}, p-value={p_value:.4f}")
```

```python
if p_value<0.05:
    print("The difference is statistically significant.")
else:
    print("The difference is not statistically significant.")
```
